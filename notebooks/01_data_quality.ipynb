## Valores nulos

```python

# Cantidad de valores nulos por columna
nulos = df.isnull().sum()

# Porcentaje de valores nulos por columna
porcentaje_nulos = (df.isnull().mean() * 100).round(2)

# Tabla resumen
resumen_nulos = pd.DataFrame({
    "Valores nulos": nulos,
    "Porcentaje de nulos (%)": porcentaje_nulos
})

# Ordenar de mayor a menor cantidad de nulos
resumen_nulos = resumen_nulos.sort_values(
    by="Valores nulos",
    ascending=False
)

print("Análisis de valores nulos por columna:")
print(resumen_nulos)
```

Es de utilidad verificar esta información en otro software ya que a veces se pueden mal intepretar cadenas vacías. O hacer este proceso más robusto aquí

## Duplicados

```python

# Valores repetidos por columna
duplicados_columna = df.apply(
    lambda columna: columna.duplicated().sum()
)

# Cantidad de valores únicos por columna
valores_unicos = df.nunique(dropna=False)

# Tabla resumen
resumen_duplicados = pd.DataFrame({
    "Valores repetidos": duplicados_columna,
    "Valores únicos": valores_unicos
})

print("Análisis de valores repetidos por columna:")
print(resumen_duplicados)
```

```python
# Total de filas completamente duplicadas (sin contar la primera aparición)
total_duplicados = df.duplicated().sum()

print("Total de filas duplicadas:", total_duplicados)
```
